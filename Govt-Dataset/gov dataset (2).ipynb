```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "MI6_GW_STW_TL.csv"

df = pd.read_csv(file_path)

df.info(), df.head()
```

```python
numerical_columns = df.select_dtypes(include=['number']).columns.tolist()
```

```python
sns.set_style("whitegrid")
```

```python
fig, axes = plt.subplots(5, 2, figsize=(15, 20))
axes = axes.flatten()
```

```python
numeric_cols = df.select_dtypes(include=['number']).columns[:2]
if len(numeric_cols) >= 2:
    plt.figure(figsize=(7, 5))
    sns.scatterplot(x=df[numeric_cols[0]], y=df[numeric_cols[1]])
    plt.title(f"Scatter Plot of {numeric_cols[0]} vs {numeric_cols[1]}")
    plt.show()
```

```python
if len(numeric_cols) > 0:
    plt.figure(figsize=(7, 5))
    sns.histplot(df[numeric_cols[0]], bins=20, kde=True)
    plt.title(f"Histogram of {numeric_cols[0]}")
    plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df.select_dtypes(include=['number']))
plt.title("Box Plot of Numeric Columns")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.violinplot(data=df.select_dtypes(include=['number']))
plt.title("Violin Plot of Numeric Columns")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.kdeplot(df[numeric_cols[0]], shade=True)
plt.title(f"KDE Plot of {numeric_cols[0]}")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
df.select_dtypes(include=['number']).plot(kind='area', alpha=0.4)
plt.title("Area Plot of Numeric Data")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.stripplot(data=df.select_dtypes(include=['number']))
plt.title("Strip Plot of Numeric Columns")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.ecdfplot(df[numeric_cols[0]])
plt.title(f"ECDF Plot of {numeric_cols[0]}")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.heatmap(df.isnull(), cmap='viridis', cbar=False)
plt.title("Missing Values Heatmap")
plt.show()
```

```python
from math import pi
num_vars = len(numeric_cols)
angles = [n / float(num_vars) * 2 * pi for n in range(num_vars)]
angles += angles[:1]
values = df[numeric_cols].mean().tolist()
values += values[:1]
plt.figure(figsize=(7, 7))
ax = plt.subplot(111, polar=True)
ax.plot(angles, values, linewidth=2, linestyle='solid')
ax.fill(angles, values, alpha=0.4)
plt.title("Radar Chart")
plt.show()
```
